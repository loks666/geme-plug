## 3. 打开插座

```python
plug.turn_on()
print("已打开")
```

## 4. 查询开关状态和功率

```python
status = plug.get_status()
print("开关状态:", "打开" if status.is_on else "关闭")

power = plug.get_power()
print("功率:", power.power, "W")
print("电流:", power.current, "A")
```
