# Geme 智能插座测试

先从下面两种连接方式中任选一个执行；两个连接 Cell 最终都会创建同一个变量 `plug`，因此后续的打开、查询、关闭和断开 Cell 可以直接复用。重复执行或切换连接方式时，会先断开已有连接。

> 当前设备网页填写订阅 `request`、发布 `response`；真机 MQTT 测试显示 SDK 需发布到 `response`、订阅 `request` 才能收到状态。结束后务必执行关闭插座和断开连接。

## 1. 连接插座

### 方式 A：不带参数，默认读取 `.env`

执行本 Cell 后直接跳到“打开插座”。

```python
from src.geme_plug import SmartPlug

previous_plug = globals().get("plug")
if previous_plug is not None:
    previous_plug.disconnect()

plug = SmartPlug()
plug.connect()
print("连接成功：使用 .env 配置")
```

### 方式 B：显式传入参数

本 Cell 显式传入 Broker、端口和主题；用户名、密码、MAC 和超时时间由 `SmartPlug` 自动从 `.env` 补齐。显式参数会覆盖 `.env` 中的同名配置。

```python
from src.geme_plug import SmartPlug

previous_plug = globals().get("plug")
if previous_plug is not None:
    previous_plug.disconnect()

plug = SmartPlug(
    host="192.168.50.70",
    port=1883,
    subscribe_topic="request",
    publish_topic="response"
)
plug.connect()
print("连接成功：使用显式参数")
```

## 2. 打开插座

```python
plug.turn_on()
print("已发送打开插座命令")
```

## 3. 查询状态和功率

```python
status = plug.get_status()
power = plug.get_power()

print("开关状态:", "打开" if status.is_on else "关闭")
print("功率:", power.power, "W")
print("电流:", power.current, "A")
```

## 4. 关闭插座

```python
plug.turn_off()
print("已发送关闭插座命令")
```

## 5. 断开连接

```python
plug.disconnect()
print("连接已断开")
```
